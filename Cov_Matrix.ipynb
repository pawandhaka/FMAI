```python
H = [1.70,1.62,1.52,1.85,1.91,1.42]
W = [72,64,84,80,72,70]
n=6
havg=0
Havg=0
Wavg=0
wavg=0
for i in range(n):
  havg=havg+H[i]
Havg=havg/n
print("averge of height",Havg)
for i in range(n):
  wavg=wavg+W[i]
Wavg=wavg/n
print("averge of weight",Wavg)
ht=0
for i in range(n):
  ht=ht+(H[i]-Havg)**2
wt=0
for i in range(n):
  wt=wt+(W[i]-Wavg)**2
varht=ht/(n-1)
print("variance of height: " ,varht)
varwt=wt/(n-1)
print("variance of weight: ",varwt)
cov=0
for i in range(n):

  cov= cov + (H[i]-Havg)*(W[i]-Wavg)
print("covariance: ",cov)
covmatrix= [[varht,cov],[cov,varwt]]
print("covariance matrix : ",covmatrix)
```
